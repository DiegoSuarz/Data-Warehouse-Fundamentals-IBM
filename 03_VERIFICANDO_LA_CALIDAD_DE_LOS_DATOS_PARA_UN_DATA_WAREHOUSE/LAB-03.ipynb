# Laboratorio Práctico: Verificando la Calidad de los Datos para un Almacén de Datos

## Propósito del Laboratorio:

El propósito principal de este laboratorio es instruir a los participantes sobre el proceso de realizar exhaustivas verificaciones de calidad de datos en un entorno de almacenamiento de datos. Se centra en el uso de un marco basado en Python dentro de una base de datos PostgreSQL para validar la integridad de los datos. Las áreas clave de énfasis incluyen la identificación de valores nulos, duplicados y entradas inválidas, así como la verificación de rangos de datos. El laboratorio tiene como objetivo equipar a los aprendices con las habilidades necesarias para configurar y utilizar un marco de pruebas para la validación de datos, asegurando la precisión y consistencia de los datos.

## Beneficios de Aprender en el Laboratorio:

Participar en este laboratorio ofrece varios beneficios, particularmente en la mejora de las capacidades en gestión de datos y aseguramiento de calidad. Los aprendices adquirirán experiencia práctica en la implementación de verificaciones automatizadas de calidad de datos, una habilidad crucial para mantener la fiabilidad de los datos en aplicaciones del mundo real. Esta competencia es especialmente beneficiosa para profesionales que trabajan con grandes conjuntos de datos, ya que asegura la integridad de los datos utilizados para el análisis y la toma de decisiones. Además, comprender estos conceptos es esencial para cualquier persona que aspire a especializarse en ciencia de datos, administración de bases de datos o cualquier campo que dependa en gran medida de datos precisos y fiables.

## Objetivos

En este laboratorio, usted:

- Verificará valores nulos
- Verificará valores duplicados
- Verificará mínimo y máximo
- Verificará valores inválidos
- Generará un informe sobre la calidad de los datos

El archivo `mytests.py` contiene todas las pruebas de calidad de datos.

Proporciona una forma rápida y fácil de crear y ejecutar nuevas pruebas de calidad de datos.

El marco de pruebas proporciona las siguientes pruebas:

- check\_for\_nulls - esta prueba verificará si hay valores nulos en una columna
- check\_for\_min\_max - esta prueba verificará si los valores en una columna están dentro de un rango de valores mínimo y máximo
- check\_for\_valid\_values - esta prueba verificará si hay valores inválidos en una columna
- check\_for\_duplicates - esta prueba verificará si hay duplicados en una columna

Cada prueba puede ser creada mencionando un mínimo de 4 parámetros.

- testname - El nombre legible por humanos de la prueba para fines de informes
- test - El nombre real de la prueba que proporciona el micro marco de pruebas
- table - El nombre de la tabla sobre la cual se realizará la prueba
- column - El nombre de la columna sobre la cual se realizará la prueba

# Ejercicio 5 - Comprobar nulos

Veamos ahora cómo es una prueba `check_for_nulls`.

Aquí hay un ejemplo de una prueba `check_for_nulls`:

```
test1={
    "testname":"Check for nulls",
    "test":check_for_nulls,
    "column": "monthid",
    "table": "DimMonth"
}
```

Todos los tests deben ser nombrados como `test` seguido de un número único para identificar el test.

- Proporciona una descripción fácil de entender para `testname`
- menciona `check_for_nulls` para `test`
- menciona el nombre de la columna en la que deseas verificar los nulos
- menciona el nombre de la tabla donde existe esta columna

Ahora creemos un nuevo test `check_for_nulls` y ejecútalo.

El test a continuación verifica si hay valores nulos en la columna `year` en la tabla `DimMonth`.

El test falla si existen nulos.

Copia y pega el código a continuación al final del archivo mytests.py.

```
test5={
    "testname":"Check for nulls",
    "test":check_for_nulls,
    "column": "year",
    "table": "DimMonth"
}
```

# Ejercicio 6 - Verificar rango mínimo y máximo

Veamos ahora cómo es una prueba `check_for_min_max`.

Aquí hay una muestra de una prueba `check_for_min_max`

```
test2={
    "testname":"Check for min and max",
    "test":check_for_min_max,
    "column": "monthid",
    "table": "DimMonth",
    "minimum":1,
    "maximum":12
}
```

Además de los campos habituales, tienes dos campos más aquí.

- minimum es el valor válido más bajo para esta columna. (Ejemplo 1 en el caso del número del mes)
- maximum es el valor válido más alto para esta columna. (Ejemplo 12 en el caso del número del mes)

Ahora vamos a crear una nueva prueba `check_for_min_max` y ejecutarla.

La prueba a continuación verifica un mínimo de 1 y un máximo de 4 en la columna `quarter` en la tabla `DimMonth`.

La prueba falla si hay valores inferiores al mínimo o superiores al máximo.

Copia y pega el código a continuación al final del archivo mytests.py.

```
test6={
    "testname":"Check for min and max",
    "test":check_for_min_max,
    "column": "quarter",
    "table": "DimMonth",
    "minimum":1,
    "maximum":4
}
```

Guarda el archivo usando Menú -\> Archivo -\> Guardar
