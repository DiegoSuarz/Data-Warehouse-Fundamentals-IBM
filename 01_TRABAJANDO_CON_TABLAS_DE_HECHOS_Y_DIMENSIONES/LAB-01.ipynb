## Propósito del laboratorio:

El laboratorio está diseñado para guiarte a través del proceso de diseño de un almacén de datos para un proveedor de servicios en la nube. Se centra en el uso de datos de facturación proporcionados en un archivo CSV para crear un esquema en estrella, incluyendo el diseño de tablas de hechos y dimensiones. Este esquema soportará consultas complejas relacionadas con la facturación, como la facturación promedio por cliente, la facturación por país, industria y categoría, así como tendencias a lo largo del tiempo.

Al completar este laboratorio, adquirirás habilidades prácticas en la organización y análisis de grandes conjuntos de datos utilizando técnicas de almacenamiento de datos. Estas habilidades son esenciales para tomar decisiones empresariales informadas, optimizar la recuperación de datos y mejorar la comprensión de las relaciones de datos. Este conocimiento es particularmente beneficioso en escenarios del mundo real, como el análisis de datos de facturación en la nube, donde puede conducir a una gestión de datos más eficiente y análisis más profundos.

# Ejercicio 2: Diseñar las tablas de hechos

El hecho en estos datos es la factura que se genera mensualmente.

Los campos `customerid` y `billedamount` son los campos importantes en la tabla de hechos.

También necesitamos una forma de identificar la información adicional del cliente, además del id, y la información de la fecha. Así que necesitamos campos que se refieran a la información del cliente y de la fecha en otras tablas.

La tabla de hechos final para la factura se vería así:

| Nombre del Campo | Detalles |
| --- | --- |
| billid | Clave primaria - Identificador único para cada factura |
| customerid | Clave foránea - Id del cliente |
| monthid | Clave foránea - Id del mes. Podemos resolver la información del mes facturado usando esto |
| billedamount | Monto cobrado por los servicios en la nube proporcionados durante ese mes en USD |

# Ejercicio 3: Diseñar las tablas de dimensiones

Hay dos dimensiones para nuestro hecho (factura mensual).

1. Información del cliente
2. Información de la fecha

Organizamos todos los campos que brindan información sobre el cliente en una tabla de dimensión.

| Nombre del Campo | Detalles |
| --- | --- |
| customerid | Clave primaria - Id del cliente |
| category | Categoría del cliente. Ejemplo: Individual o Empresa |
| country | País del cliente |
| industry | Dominio/industria al que pertenece el cliente. Ejemplo: Legal, Ingeniería |

Organizamos o derivamos todos los campos que brindan información sobre la fecha de la factura.

| Nombre del Campo | Detalles |
| --- | --- |
| monthid | Clave primaria - Id del mes |
| year | Año derivado del campo de mes de los datos originales. Ejemplo: 2010 |
| month | Número del mes derivado del campo de mes de los datos originales. Ejemplo: 1, 2, 3 |
| monthname | Nombre del mes derivado del campo de mes de los datos originales. Ejemplo: Marzo |
| quarter | Número del trimestre derivado del campo de mes de los datos originales. Ejemplo: 1, 2, 3, 4 |
| quartername | Nombre del trimestre derivado del campo de mes de los datos originales. Ejemplo: T1, T2, T3, T4 |

# CREAR LA BASE DE DATOS

````sql
Create database customer;
````

````sql
CREATE TABLE public."FactBilling"
(
    rowid integer NOT NULL,
    customerid integer NOT NULL,
    monthid integer NOT NULL,
    billedamount integer NOT NULL,
    PRIMARY KEY (rowid)
);
````

````sql
CREATE TABLE public."DimMonth"
(
    monthid integer NOT NULL,
    year integer NOT NULL,
    month integer NOT NULL,
    monthname "char" NOT NULL,
    quarter integer NOT NULL,
    quartername "char" NOT NULL,
    PRIMARY KEY (monthid)
);
````

````sql
CREATE TABLE public."DimCustomer"
(
    customerid integer NOT NULL,
    category "char" NOT NULL,
    country "char" NOT NULL,
    industry "char" NOT NULL,
    PRIMARY KEY (customerid)
);
````

````sql
SELECT schemaname, tablename
FROM pg_catalog.pg_tables
WHERE schemaname = 'public'
ORDER BY schemaname, tablename;
````

# CREAR LAS CLAVES FORANEAS:

````sql
ALTER TABLE public."FactBilling"
    ADD FOREIGN KEY (customerid)
    REFERENCES public."DimCustomer" (customerid)
    NOT VALID;


ALTER TABLE public."FactBilling"
    ADD FOREIGN KEY (monthid)
    REFERENCES public."DimMonth" (monthid)
    NOT VALID;
````

````sql
SELECT
    tc.table_schema,
    tc.table_name,
    kc.column_name
FROM information_schema.table_constraints AS tc
JOIN information_schema.key_column_usage AS kc
  ON kc.constraint_name = tc.constraint_name
WHERE tc.constraint_type = 'PRIMARY KEY'
  AND tc.table_name = 'FactBilling';
````

````sql
SELECT
    tc.table_name AS tabla_hija,
    kcu.column_name AS columna_hija,
    ccu.table_name AS tabla_padre,
    ccu.column_name AS columna_padre
FROM
    information_schema.table_constraints AS tc
JOIN information_schema.key_column_usage AS kcu
    ON tc.constraint_name = kcu.constraint_name
JOIN information_schema.constraint_column_usage AS ccu
    ON ccu.constraint_name = tc.constraint_name
WHERE
    constraint_type = 'FOREIGN KEY'
    AND tc.table_name = 'FactBilling';
````

````sql
SELECT
    tablename,
    indexname,
    indexdef
FROM
    pg_indexes
WHERE
    schemaname = 'public'
    AND tablename = 'FactBilling';
````
