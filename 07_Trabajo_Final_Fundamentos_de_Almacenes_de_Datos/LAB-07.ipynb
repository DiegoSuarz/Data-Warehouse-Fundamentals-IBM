# Trabajo Final: Fundamentos de Almacenes de Datos

Este laboratorio integral está diseñado para proporcionar experiencia práctica en el diseño, implementación y consulta de un almacén de datos utilizando PostgreSQL. Simula un escenario del mundo real donde tú, como ingeniero de datos, asistes a una empresa de gestión de residuos en Brasil en la gestión y análisis de sus datos de recolección de residuos sólidos. El laboratorio involucra múltiples etapas, desde el diseño y creación de un esquema en estrella para el almacén de datos, la carga de datos, la escritura de consultas SQL complejas para la agregación de datos y la creación de vistas materializadas para la optimización de consultas.  

#### **Lo que aprenderás:**

El laboratorio ofrece una multitud de beneficios de aprendizaje, particularmente para aquellos que buscan mejorar sus habilidades en ingeniería de datos e inteligencia empresarial:

- **Carga y transformación de datos**: Ofrece práctica en la carga y transformación de datos, una habilidad esencial para gestionar almacenes de datos.
    
- **Aplicaciones en escenarios del mundo real**: El enfoque basado en escenarios del laboratorio asegura que las habilidades adquiridas sean relevantes y aplicables en proyectos reales de almacenamiento de datos e inteligencia empresarial.
    
- **Avance profesional**: Estas habilidades están en alta demanda en los campos de la ingeniería de datos, inteligencia empresarial y análisis, contribuyendo significativamente al crecimiento profesional y las oportunidades.
    

Este laboratorio sirve como una guía integral para cualquiera que aspire a fortalecer su experiencia en almacenamiento de datos e inteligencia empresarial, proporcionando habilidades prácticas que son directamente aplicables en entornos profesionales.

# Escenario

Eres un ingeniero de datos contratado por una empresa de gestión de residuos sólidos. Recoge y recicla residuos sólidos en las principales ciudades del país de Brasil. Operan cientos de camiones de diferentes tipos para recoger y transportar residuos sólidos. La empresa desea crear un almacén de datos para poder generar informes como:

- Total de residuos recogidos por año por ciudad
- Total de residuos recogidos por mes por ciudad
- Total de residuos recogidos por trimestre por ciudad
- Total de residuos recogidos por año por tipo de camión
- Total de residuos recogidos por tipo de camión por ciudad
- Total de residuos recogidos por tipo de camión por estación por ciudad

Utilizarás tus habilidades en almacenes de datos para diseñar e implementar un almacén de datos para la empresa.

#### Objetivos de aprendizaje

Después de completar este laboratorio, podrás:

- Diseñar un almacén de datos.
- Cargar datos en el almacén de datos.
- Crear una vista materializada.
