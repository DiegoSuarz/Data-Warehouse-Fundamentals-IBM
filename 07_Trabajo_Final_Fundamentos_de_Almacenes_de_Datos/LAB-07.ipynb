#### Tarea 13: Crear una consulta de conjuntos de agrupamiento

Crea una consulta de conjuntos de agrupamiento utilizando las columnas stationid, trucktype y total waste collected.

Toma una captura de pantalla del SQL y de las filas de salida.

Nombra la captura de pantalla `13-groupingsets.jpg`. (Las imágenes se pueden guardar con la extensión .jpg o .png.)

```sql
 SELECT  s.stationid, t.trucktype, SUM(f.wastecollected) AS "Total waste"
 FROM "FactTrips" f
 INNER JOIN "DimStation" s
 ON f.stationid = s.stationid
 INNER JOIN "DimTruck" t
 ON f.truckid = t.truckid
 GROUP BY  GROUPING SETS((s.stationid, t.trucktype),
                        s.stationid, t.trucktype)
ORDER BY s.stationid, t.trucktype
;
```

<span style="color: var(--vscode-foreground);">Tarea 14: Crear una consulta de rollup</span>  

Crea una consulta de rollup utilizando las columnas year, city, stationid y total waste collected.

Toma una captura de pantalla del SQL y de las filas de salida.

Nombra la captura de pantalla `14-rollup.jpg`. (Las imágenes se pueden guardar con la extensión .jpg o .png.)

```sql
SELECT  d.year, s.city, s.stationid, SUM(wastecollected)
FROM "FactTrips" f 
INNER JOIN "DimDate" d 
ON f.dateid = d.dateid
INNER JOIN "DimStation" s 
ON f.stationid = s.stationid
GROUP BY ROLLUP(d.year, s.city, s.stationid)
ORDER BY d.year, s.city, s.stationid
;
```

#### Tarea 15: Crear una consulta de cube

Crea una consulta de cube utilizando las columnas year, city, stationid y average waste collected.

Toma una captura de pantalla del SQL y de las filas de salida.

Nombra la captura de pantalla `15-cube.jpg`. (Las imágenes se pueden guardar con la extensión .jpg o .png.)

```sql
SELECT d.year, s.city, s.stationid, ROUND(AVG(wastecollected),2) as Average
FROM "FactTrips" f 
INNER JOIN "DimDate" d 
ON f.dateid = d.dateid
INNER JOIN "DimStation" s
ON f.stationid = s.stationid
GROUP BY CUBE(d.year, s.city, s.stationid)
ORDER BY  d.year, s.city, s.stationid;
```

#### <span style="color: var(--vscode-foreground);">Tarea 16: Crear una vista materializada</span>  

Crea una vista materializada llamada max\_waste\_stats utilizando las columnas city, stationid, trucktype y max waste collected.

Toma una captura de pantalla del SQL.

Nombra la captura de pantalla `16-mv.jpg`. (Las imágenes se pueden guardar con la extensión .jpg o .png.)

```sql
DROP MATERIALIZED VIEW max_waste_stats;
```

```sql
CREATE MATERIALIZED VIEW max_waste_stats AS
SELECT s.city, s.stationid, t.trucktype, MAX(f.wastecollected) as "Max waste collected" 
FROM "FactTrips" AS f 
INNER JOIN "DimStation" As s
ON f.stationid = s.stationid
INNER JOIN "DimTruck" AS t
ON f.truckid = t.truckid
GROUP BY s.city, s.stationid, t.trucktype
ORDER BY s.city, s.stationid, t.trucktype
WITH DATA
```

```sql
REFRESH MATERIALIZED VIEW max_waste_stats;
```

```sql
SELECT * FROM max_waste_stats;
```
