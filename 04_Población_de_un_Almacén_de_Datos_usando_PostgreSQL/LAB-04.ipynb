# [Laboratorio Práctico: Población de un Almacén de Datos usando PostgreSQL](https://cf-courses-data.static.labs.skills.network/IBM-DB0260EN-SkillsNetwork/labs/BIWorkaroundFiles/week2/Populating_datawarehouse_with_postgres.md.html)

## Propósito del Laboratorio:

El laboratorio está diseñado para proporcionar experiencia práctica en la creación y gestión de una base de datos de producción utilizando PostgreSQL dentro del IBM Skills Network Labs (SN Labs) Cloud IDE. Aprenderás a lanzar una instancia de servidor PostgreSQL, utilizar la interfaz gráfica de usuario (GUI) pgAdmin para operaciones de base de datos y ejecutar tareas esenciales como crear una base de datos, diseñar tablas y cargar datos. El laboratorio se centra en construir una base en la gestión de bases de datos guiando a los aprendices a través del proceso de configurar una base de datos de ‘Producción’ y poblarla con datos siguiendo un diseño de esquema en estrella.

## Beneficios de Aprender en el Laboratorio:

Participar en este laboratorio ofrece beneficios significativos para los aprendices que buscan profundizar su comprensión de los sistemas de gestión de bases de datos, particularmente PostgreSQL. Al trabajar en el laboratorio, adquirirás habilidades prácticas en SQL, creación de bases de datos, diseño de tablas y manipulación de datos, que son cruciales para roles en ingeniería de datos, administración de bases de datos y ciencia de datos. El enfoque práctico ayuda a consolidar el conocimiento de esquemas de bases de datos y consultas SQL, mejorando así la capacidad del aprendiz para gestionar y analizar datos de manera efectiva en escenarios del mundo real. Además, la familiaridad con herramientas como pgAdmin y el entorno Cloud IDE añade experiencia valiosa a tu conjunto de habilidades, preparándote para proyectos y tareas avanzadas de bases de datos.

#### Problema 2: Usando la herramienta PostgreSQL, crea una vista materializada simple llamada avg\_customer\_bill con los campos customerid y averagebillamount.

<details><summary>Pista (Haz clic aquí) </summary>

Usa el comando crear vista materializada.

</details>
<br>
<details><summary>Solución (Haz clic aquí) </summary>

```
CREATE MATERIALIZED VIEW  avg_customer_bill (customerid, averagebillamount) AS
(select customerid, avg(billedamount)
from public."FactBilling"
group by customerid
);
```

Haz clic en el botón **Ejecutar Todo** para ejecutar la declaración. Deberías ver el estado como **Éxito** en la sección de **Resultado**.

</details>

#### Problema 3: Actualiza las vistas materializadas recién creadas

<details><summary>Pista (Haz clic aquí) </summary>

Utiliza el comando de actualización de vista materializada.

</details>
<br>
<details><summary>Solución (Haz clic aquí) </summary>

```
REFRESH MATERIALIZED VIEW avg_customer_bill;
```
</details>

#### Problema 4: Usando las vistas materializadas recién creadas, encuentra los clientes cuyo promedio de facturación es superior a 11000.

<details><summary>Pista (Haz clic aquí) </summary>

Utiliza la sentencia select en las vistas materializadas con una cláusula where en la columna averagebillamount.

</details>
<br>
<details><summary>Solución (Haz clic aquí) </summary>

```
select * from avg_customer_bill where averagebillamount > 11000;
```
</details>
