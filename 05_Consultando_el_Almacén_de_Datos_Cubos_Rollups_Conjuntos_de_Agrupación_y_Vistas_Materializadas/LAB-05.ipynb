**GROUPING SETS, CUBE y ROLLUP** nos permiten crear fácilmente subtotales y totales generales de diversas maneras. Todos estos operadores se utilizan junto con el operador GROUP BY.

El operador **GROUPING SETS** nos permite agrupar datos de diferentes maneras en una única declaración SELECT.

El operador **ROLLUP** se utiliza para crear subtotales y totales generales para un conjunto de columnas. Los totales resumidos se crean en función de las columnas pasadas al operador ROLLUP.

El operador **CUBE** produce subtotales y totales generales. Además, produce subtotales y totales generales para cada permutación de las columnas proporcionadas al operador CUBE.

# Ejercicio 2 - Escribe una consulta utilizando conjuntos de agrupamiento

select year,category, sum(billedamount) as totalbilledamount

# Ejercicio 5 - Crear una Tabla de Consulta Materializada (Vistas materializadas)

En pgAdmin podemos implementar vistas materializadas utilizando Tablas de Consulta Materializadas.

#### Paso 1: Crear las vistas materializadas.

Ejecute la instrucción sql a continuación para crear una vista materializada llamada countrystats.

```
CREATE MATERIALIZED VIEW countrystats (country, year, totalbilledamount) AS
(select country, year, sum(billedamount)
from "FactBilling"
left join "DimCustomer"
on "FactBilling".customerid = "DimCustomer".customerid
left join "DimMonth"
on "FactBilling".monthid="DimMonth".monthid
group by country,year);
```

El comando anterior crea una vista materializada llamada countrystats que tiene 3 columnas.

- País
- Año
- totalbilledamount

La vista materializada es esencialmente el resultado de la consulta a continuación, que te proporciona el año, el nombre del trimestre y la suma del monto facturado agrupada por año y nombre del trimestre.

```
select year, quartername, sum(billedamount) as totalbilledamount
from "FactBilling"
left join "DimCustomer"
on "FactBilling".customerid = "DimCustomer".customerid
left join "DimMonth"
on "FactBilling".monthid="DimMonth".monthid
group by grouping sets(year, quartername);
```

#### Paso 2: Población/actualización de datos en las vistas materializadas.

Ejecute la instrucción sql a continuación para poblar las vistas materializadas countrystats.

```
REFRESH MATERIALIZED VIEW countrystats;
```

El comando anterior llena las vistas materializadas con datos relevantes.

Paso 3: Consulta las vistas materializadas.

Una vez que una vista materializada se actualiza, puedes consultarla.

Ejecuta la declaración sql a continuación para consultar las vistas materializadas countrystats.

```
select * from countrystats;
```

<span style="color: rgb(232, 230, 227); font-family: Arial; background-color: rgb(26, 28, 29);">Crear un CUBE utilizando las tres columnas etiquetadas como año, categoría y suma de billedamount</span>

````sql
SELECT DM.year, DC.category, SUM(FB.billedamount) AS totalbilledamount
FROM "FactBilling" AS FB
inner join "DimMonth" AS DM
ON FB.monthid = DM.monthid
inner join "DimCustomer" AS DC
ON FB.customerid = DC.customerid
GROUP BY CUBE(DM.year, DC.category)
--ORDER BY DM.year, DC.category
;
````

````sql
CREATE MATERIALIZED VIEW average_billamount(year,quarter,category,country,PromedioBilledamount )
AS
(
SELECT DM.year, DM.quarter, DC.category, DC.country, AVG(FB.billedamount) as PromedioBilledamount
FROM "FactBilling" AS FB
INNER JOIN "DimMonth" AS DM
ON FB.monthid = DM.monthid
INNER JOIN "DimCustomer" AS DC
ON FB.customerid = DC.customerid
GROUP BY DM.year, DM.quarter, DC.category, DC.country
);

````
