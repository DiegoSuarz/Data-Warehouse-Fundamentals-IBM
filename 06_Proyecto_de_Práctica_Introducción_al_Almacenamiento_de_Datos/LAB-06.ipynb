## Tarea 13: Crea una consulta de conjuntos de agrupamiento  

Crea una consulta de conjuntos de agrupamiento (set grouping) utilizando las columnas productid, producttype, total sales.  

Toma una captura de pantalla del SQL y de las filas de salida.  

Nombra la captura de pantalla <span style="color: #ce9178;">`13-groupingsets.jpg`</span>. (Las imágenes se pueden guardar con la extensión .jpg o .png.)

<details><summary>Solución (Haz clic aquí) </summary>

  

```
SELECT

    p.Productid,

    p.Producttype,

    SUM(f.Price\_PerUnit \* f.QuantitySold) AS TotalSales

FROM

    FactSales f

INNER JOIN

    DimProduct p ON f.Productid = p.Productid

GROUP BY GROUPING SETS (

    (p.Productid, p.Producttype),

    p.Productid,

    p.Producttype,

    ()

)

ORDER BY

    p.Productid,

    p.Producttype;

```


En esta consulta:

\- Estás uniendo FactSales ‘f’ con DimProduct ‘p’ en su productid para correlacionar cada venta con su tipo de producto.

\- Estás utilizando GROUPING SETS para especificar los diferentes niveles de agregación:

  

– Por ambas Productid y Producttype  

– Solo por Productid  

– Solo por Producttype  

– Y un total general con (), que no agrupa por ninguna columna y, por lo tanto, devuelve la suma de todas las ventas.
 

\- Estás calculando TotalSales multiplicando el Price_PerUnit por QuantitySold para cada venta.

  
La cláusula ORDER BY asegura que los resultados estén ordenados por productid y luego por producttype.

</details>
<br>

````sql
/*Crea una consulta de conjuntos de agrupamiento (set grouping) utilizando 
las columnas productid, productname, total sales.  */

SELECT DP.productid, DP.productname, SUM(FS.quantitysold * FS.priceperunit) as "Total Sales"
FROM "MyfactSales" AS FS
INNER JOIN "MyDimProduct" AS DP
ON FS.productid = DP.productid
GROUP BY GROUPING SETS(
    (DP.productid, DP.productname),
    DP.productid, DP.productname
)
ORDER BY DP.productid, DP.productname
;
````

Crea una consulta rollup utilizando las columnas año, ciudad, productid y total de ventas. Toma una captura de pantalla del SQL y de las filas de salida.

Nombra la captura de pantalla `14-rollup.jpg`. (Las imágenes se pueden guardar con la extensión .jpg o .png.)

<details><summary>Solución (Haz clic aquí) </summary>

```
SELECT
    d.Year,
    cs.City,
    p.Productid,
    SUM(f.Price_PerUnit * f.QuantitySold) AS TotalSales 
FROM
    FactSales f
JOIN
    DimDate d ON f.Dateid = d.Dateid
JOIN
    DimProduct p ON f.Productid = p.Productid
JOIN
    DimCustomerSegment cs ON f.Segmentid = cs.Segmentid
GROUP BY ROLLUP (d.Year, cs.City, p.Productid)
ORDER BY
    d.Year DESC,
    cs.City,
    p.Productid;

```

Esta consulta realiza las siguientes operaciones:

- Une FactSales con DimDate en Dateid, DimProduct en Productid y DimCustomerSegment en Segmentid.
- Selecciona el año de DimDate, la ciudad de DimCustomerSegment y el ID del producto de DimProduct.
- Calcula las ventas totales multiplicando el precio por unidad por la cantidad vendida para cada entrada de venta.
- Agrupa los resultados utilizando la función ROLLUP para crear un conjunto de agrupamiento que incluye todas las combinaciones de año, ciudad y productid, junto con sus respectivos subtotales y un total general para todas las ventas.

La cláusula ORDER BY asegura que los resultados se ordenen primero por año en orden descendente, luego por ciudad y ID del producto

</details>
<br>

````sql
SELECT DD.year, DC.segmentname, DP.productid, SUM(FS.quantitysold * FS.priceperunit) AS "Total Ventas"
FROM "MyfactSales" AS FS
INNER JOIN "MyDimDate" AS DD
ON FS.dateid = DD.dateid
INNER JOIN "MyDimCustomerSegment" AS DC
ON FS.segmentid = DC.segmentid
INNER JOIN "MyDimProduct" AS DP
ON FS.productid = DP.productid
GROUP BY ROLLUP (DD.year, DC.segmentname, DP.productid)
ORDER BY DD.year DESC, DC.segmentname, DP.productid
;
````

### Tarea 15: Crear una consulta de cubo

Crea una consulta de cubo utilizando las columnas año, ciudad, productid y ventas promedio.  
Toma una captura de pantalla del SQL y de las filas de salida.  
Nombra la captura de pantalla `15-cube.jpg`. (Las imágenes se pueden guardar con la extensión .jpg o .png.)

<details><summary>Solución (Haz clic aquí) </summary>

```
SELECT
    d.Year,
    cs.City,
    p.Productid,
    AVG(f.Price_PerUnit * f.QuantitySold) AS AverageSales
FROM
    FactSales f
INNER JOIN
    DimDate d ON f.Dateid = d.Dateid
INNER JOIN
    DimProduct p ON f.Productid = p.Productid
INNER JOIN
    DimCustomerSegment cs ON f.Segmentid = cs.Segmentid
GROUP BY CUBE (d.Year, cs.City, p.Productid);

```

En esta consulta:

- La cláusula CUBE se utiliza en el GROUP BY para crear subtotales para todas las combinaciones de año, ciudad y productid, además del total general de todos los grupos.
    
- AVG(f.Price\_PerUnit \* f.QuantitySold) calcula el promedio de ventas, teniendo en cuenta tanto el precio por unidad como la cantidad vendida.

</details>

<br>

````sql
SELECT DD.year,  DC.segmentname, DP.productid, ROUND(AVG(FS.quantitysold * FS.priceperunit),2) as "Promedio de ventas"
FROM "MyfactSales" AS FS
INNER JOIN "MyDimDate" AS DD
ON FS.dateid  = DD.dateid
INNER JOIN "MyDimCustomerSegment" AS DC
ON FS.segmentid = DC.segmentid
INNER JOIN "MyDimProduct" AS DP
ON FS.productid = DP.productid
GROUP BY CUBE(DD.year,  DC.segmentname, DP.productid)
````

# Tarea 16: Crear una vista materializada

Crea una vista materializada llamada max\_sales utilizando las columnas ciudad, productid, producttype y max sales.  
Toma una captura de pantalla del SQL.  
Nombra la `screenshot 16-mv.jpg`. (Las imágenes se pueden guardar con la extensión .jpg o .png.)

<details><summary>Solución (Haz clic aquí) </summary>

```
CREATE MATERIALIZED VIEW max_sales AS
SELECT
    cs.City,
    p.Productid,
    p.Producttype,
    MAX(f.Price_PerUnit * f.QuantitySold) AS MaxSales
FROM
    FactSales f
JOIN
    DimProduct p ON f.Productid = p.Productid
JOIN
    DimCustomerSegment cs ON f.Segmentid = cs.Segmentid
GROUP BY
    cs.City,
    p.Productid,
    p.Producttype
WITH DATA;

```

Esta declaración creará la vista materializada y la poblará con los datos actuales de las tablas unidas. La cláusula WITH DATA le dice a PostgreSQL que llene la vista con los resultados de la consulta de inmediato. Si quisieras crear la vista sin llenarla de datos, usarías WITH NO DATA.

Para actualizar la vista materializada con los datos más recientes, usarías el comando REFRESH MATERIALIZED VIEW:

`REFRESH MATERIALIZED VIEW max_sales;`

</details>

<br>


````sql
CREATE MATERIALIZED VIEW max_sales AS

SELECT CS.segmentname, DP.productid, DP.productname, MAX(FS.quantitysold * FS.priceperunit) AS "Venta Maxima"
FROM "MyfactSales" AS FS
INNER JOIN "MyDimCustomerSegment" AS CS
ON FS.segmentid = CS.segmentid
INNER JOIN "MyDimProduct" AS DP
ON FS.productid = DP.productid 
GROUP BY CS.segmentname, DP.productid, DP.productname
WITH DATA;
````

````sql
REFRESH MATERIALIZED VIEW max_sales;
````
