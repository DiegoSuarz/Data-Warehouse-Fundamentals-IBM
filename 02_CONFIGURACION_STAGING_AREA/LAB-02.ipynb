# Laboratorio Práctico: Configuración de un área de preparación

## Propósito del Laboratorio:

El propósito del laboratorio es equiparte con habilidades prácticas en la configuración y gestión de un servidor de preparación para un almacén de datos, específicamente utilizando PostgreSQL. El laboratorio se centra en enseñar cómo diseñar e implementar un esquema de base de datos, cargar datos en tablas y ejecutar consultas de muestra para interactuar con los datos. Esto tiene como objetivo proporcionarte una comprensión práctica de las complejidades involucradas en la preparación y gestión de un entorno de almacén de datos.

## Beneficios de Aprender en el Laboratorio:

El principal beneficio de este laboratorio es la experiencia práctica que ofrece en conceptos y prácticas de almacenamiento de datos. Al trabajar en el Skills Network Cloud IDE, un entorno interactivo, puedes practicar y perfeccionar tus habilidades en la gestión de bases de datos. Esta exposición práctica es crucial para desarrollar una base sólida en la gestión y análisis de datos, habilidades que son altamente valoradas en escenarios empresariales del mundo real. El laboratorio te ayuda a comprender los aspectos operativos de los almacenes de datos, preparándote mejor para los desafíos en la gestión y análisis de datos en entornos profesionales.

## Objetivos

En este laboratorio tú:

- Configurarás un servidor de preparación para un almacén de datos
- Crearás el esquema para almacenar los datos
- Cargarás los datos en las tablas
- Ejecutarás una consulta de muestra

# Ejercicio 2 - Crear Base de Datos

Crea la base de datos en el almacén de datos.

Ahora, abre una nueva terminal, haciendo clic en la barra de menú y seleccionando Terminal-\>Nueva Terminal. Esto abrirá una nueva terminal en la parte inferior de la pantalla.

Ejecuta los comandos a continuación en la terminal recién abierta. (Puedes copiar el código haciendo clic en el pequeño botón de copiar en la esquina inferior derecha del bloque de código a continuación y luego pegarlo donde desees).

Usando el comando createdb del servidor PostgreSQL, podemos crear directamente la base de datos desde la terminal. Antes de eso, exporta tu contraseña reemplazando \<your\_password\> con tu contraseña del servidor PostgreSQL (que se puede encontrar en la pestaña de información de conexión) en el siguiente comando y ejecútalo.

```
export PGPASSWORD=<your_password>;
```

Luego ejecuta el siguiente comando que creará una base de datos llamada billingDW.

```
createdb -h postgres -U postgres -p 5432 billingDW
```

En el comando anterior

- `-h` menciona que el servidor de la base de datos es accesible utilizando el nombre de host “postgres”
- `-U` menciona que estamos usando el nombre de usuario postgres para iniciar sesión en la base de datos
- `-p` menciona que el servidor de la base de datos se está ejecutando en el puerto número 5432

Deberías ver una salida como esta.

# Ejercicio 4 - Cargar datos en tablas de Dimensión

Cuando cargamos datos en las tablas, es una buena práctica cargar primero los datos en las tablas de dimensión.

Paso 1: Cargar datos en la tabla DimCustomer

Ejecute el comando a continuación para cargar los datos en la tabla DimCustomer en la base de datos `billingDW`.

```
psql  -h postgres -U postgres -p 5432 billingDW < DimCustomer.sql
```

Paso 2: Cargar datos en la tabla DimMonth

Ejecute el comando a continuación para cargar los datos en la tabla DimMonth en la base de datos `billingDW`.

```
psql  -h postgres -U postgres -p 5432 billingDW < DimMonth.sql
```

# Ejercicio 5 - Cargar datos en la tabla Fact

Cargar datos en la tabla FactBilling

Ejecute el siguiente comando para cargar los datos en la tabla FactBilling en la base de datos `billingDW`.
